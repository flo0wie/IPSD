## 10. Conditionals

`if`, `elif`, dan `else` digunakan untuk membuat keputusan berdasarkan kondisi.

Kita akan membuat kategori performa sederhana.

````python
def performance_category(score):
    if score >= 85:
        return "Excellent"
    elif score >= 70:
        return "Good"
    else:
        return "Needs Support"

df["performance"] = df["quiz_score"].apply(performance_category)
df[["student_id", "quiz_score", "performance"]]
````

## 11. Functions

Function membuat kode dapat digunakan kembali.

Buat fungsi untuk menghitung **Study Engagement Score** sederhana:

> `0.4 × attendance + 0.3 × assignment_completion_rate + 0.3 × quiz_score`

Assignment completion rate dihitung dari 10 tugas.

````python
def engagement_score(attendance_pct, assignments_done, quiz):
    assignment_rate = assignments_done / 10 * 100
    return (
        0.4 * attendance_pct
        + 0.3 * assignment_rate
        + 0.3 * quiz
    )

df["engagement_score"] = df.apply(
    lambda row: engagement_score(
        row["attendance"],
        row["assignments_completed"],
        row["quiz_score"]
    ),
    axis=1
)

display(df[["student_id", "engagement_score"]])
````

## 12. Exceptions

Program dapat mengalami error. `try-except` digunakan untuk menangani kondisi error secara terkontrol.

Contoh kasus: input jam belajar harus berupa angka.

````python
def convert_study_hours(value):
    try:
        return float(value)
    except (ValueError, TypeError):
        return None

print(convert_study_hours("4.5"))
print(convert_study_hours("empat"))
````

## 13. Class dan Object

`class` digunakan sebagai blueprint untuk membuat object.

Dalam studi kasus ini, kita membuat class `Student`.

````python
class Student:
    def __init__(self, student_id, study_hours, attendance, quiz_score):
        self.student_id = student_id
        self.study_hours = study_hours
        self.attendance = attendance
        self.quiz_score = quiz_score

    def study_status(self):
        if self.study_hours >= 4 and self.quiz_score >= 70:
            return "On Track"
        return "Needs Review"

    def summary(self):
        return f"{self.student_id}: {self.study_hours} jam belajar, nilai {self.quiz_score}"

student_obj = Student("M010", 5, 90, 82)

print(student_obj.summary())
print(student_obj.study_status())
````

## 15. Generators

Generator menghasilkan data secara bertahap menggunakan `yield`. Ini berguna ketika data sangat besar sehingga kita tidak ingin membuat seluruh hasil sekaligus di memori.

````python
def score_generator(scores):
    for score in scores:
        yield score

gen = score_generator(df["quiz_score"])

for score in gen:
    print("Score:", score)
````

## 16. NumPy untuk Analisis Numerik

Kita dapat menghitung statistik dan hubungan sederhana menggunakan NumPy.

````python
hours = df["study_hours"].to_numpy()
scores = df["quiz_score"].to_numpy()

print("Mean jam belajar :", np.mean(hours))
print("Median jam belajar:", np.median(hours))
print("Std jam belajar  :", np.std(hours))
print("Korelasi sederhana jam belajar vs nilai:", np.corrcoef(hours, scores)[0, 1])
````

## 17. Menggabungkan Semua Konsep: Mini Data Science Pipeline

Sekarang kita gabungkan:

**Input → Validation → Transformation → Analysis → Decision**

Pipeline berikut:
1. Membaca data.
2. Memastikan kolom penting tersedia.
3. Mengubah data menjadi numerik.
4. Menghitung engagement score.
5. Memberikan kategori risiko.
6. Menghasilkan ringkasan.

````python
required_columns = [
    "student_id",
    "study_hours",
    "attendance",
    "assignments_completed",
    "quiz_score"
]

def validate_columns(dataframe, columns):
    missing = [column for column in columns if column not in dataframe.columns]
    if missing:
        raise ValueError(f"Kolom belum tersedia: {missing}")
    return True

def calculate_risk(row):
    if row["quiz_score"] < 60 or row["attendance"] < 70:
        return "High"
    elif row["quiz_score"] < 75 or row["attendance"] < 80:
        return "Medium"
    else:
        return "Low"

try:
    validate_columns(df, required_columns)

    numeric_columns = [
        "study_hours",
        "attendance",
        "assignments_completed",
        "quiz_score"
    ]

    for col in numeric_columns:
        df[col] = pd.to_numeric(df[col], errors="coerce")

    df["risk_level"] = df.apply(calculate_risk, axis=1)

    print("Pipeline berhasil dijalankan.")
    display(df)

except ValueError as error:
    print("Validation error:", error)
````
