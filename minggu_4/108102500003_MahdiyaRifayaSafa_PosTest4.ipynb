```python
kolom = ['alcohol', 'volatile acidity', 'sulphates']

statistik = df_analisis[kolom].agg(['mean', 'median', 'std']).T

statistik.columns = ['mean', 'median', 'simpangan baku']

statistik
```

```python
statistik = df_analisis.select_dtypes(include='number').drop(columns='quality').agg(['mean', 'median']).T

statistik['selisih'] = (statistik['mean'] - statistik['median']).abs()

statistik = statistik.sort_values('selisih', ascending=False)

statistik
```

variabel total sulfur dioxide punya selisih mean n mediang paling besar dibandingin variabel lainnya. kemungkinan disebabkan oleh distribusi data yg menceng ke kanan [karena mean lebih besar dri median], yaitu adanya bbrapa sampel paling dgn nilai sulfur dioxide yg jauh lebih tinggi dibandingkan sebagian besar data. nilai sulfur dioxide yang jauh lebih tinggi dibandingkan sebagian besar data. Nilai tinggi tersebut dapat memengaruhi mean sehingga nilainya menjadi lebih besar daripada median. oleh karena ituuuuuu, median lebih mewakili pemusatan data karena tidak mudah terpengaruh oleh nilai ekstrem

mean plot hanya menunjukkan nilai rata rata setiap kelompok, shg tdk memperlihatkan sebaran data, median, dan nilai ekstrem. dua kelompok bisa memiliki mean yg mirip, tetapi sebaran datanya sangat berbeda. karea itu, box plot dapat membantu memahami distribusi antarkelompok dengan lebih lengkap

```python
def kelompok_quality(nilai):
    if nilai <= 4:
        return 'Rendah'
    elif nilai <= 6:
        return 'Sedang'
    else:
        return 'Tinggi'

df_analisis['quality_class'] = df_analisis['quality'].apply(kelompok_quality)

df_analisis[['quality', 'quality_class']].head()
```

```python
df_analisis['quality_class'].value_counts().sort_index()
```

pengelompokan ini membuat hasil lebih sederhana dan lebih mudah dibandingkan, tetapi informasi detail berkurang karena nilai quality yang berbeda digabungkan ke dalam kelas yang sama.
