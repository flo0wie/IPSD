`````python
scaler = StandardScaler()
scaled_house_price = scaler.fit_transform(house_price)
scaled_house_price
`````

`````python
scaled_house_price=pd.DataFrame(scaled_house_price, columns=house_price.columns)
scaled_house_price
`````

`````python
x = scaled_house_price.drop("SalePrice", axis=1).values
y = scaled_house_price["SalePrice"].values.reshape(-1,1)

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.3, random_state=0)
`````

`````python
print("Ukuran x train", len(x_train))
print("Ukuran x test", len(x_test))
print("Ukuran y train", len(x_train))
print("Ukuran y test", len(x_test))
`````

$$ Y=\beta_0+\beta_1X_1+\beta_2X_2+\beta_3X_3+\epsilon$$

`````python
input = sm.add_constant(x_train)
`````

`````python
SimpleModel = sm.OLS(y_train, input, missing = "drop")
results = SimpleModel.fit()
`````

`````python
print(results.summary())
`````

`````python
input_test = sm.add_constant(x_test)
prediction_1 = results.predict(input_test)

list_columns = []
for i in house_price.columns:
    if i !="SalePrice":
        list_columns.append(i)
list_columns

df_statmod = pd.DataFrame(x_test, columns=list_columns)
df_statmod ["actual_price"] = y_test
df_statmod ["prediction_price"] = prediction_1
df_statmod.head()
`````

`````python
sns.kdeplot(df_statmod['actual_price'], label="Actual", fill=True)
sns.kdeplot(df_statmod['prediction_price'], label="Prediction", fill=True)

plt.legend()
plt.show()
`````

$$ MSE (Mean Square Error) : \frac{1}\sum_i(y_1-\hat{y}_i)^2 $$

$$ MAE (Mean Absolut Error) : \frac{1}{n}\sum_i |y_i-\hat{y}_i|$$

$$ RMSE = \sqrt{MSE}$$

`````python
print("R-square:", metrics.r2_score(df_statmod['actual_price'],
                                    df_statmod['prediction_price']))
`````

`````python
print("MSE:", metrics.mean_squared_error(df_statmod['actual_price'],
                                    df_statmod['prediction_price']))
`````

`````python
print("MAE:", metrics.mean_absolute_error(df_statmod['actual_price'],
                                    df_statmod['prediction_price']))
`````

`````python
print("RMSE:", np.sqrt(metrics.mean_squared_error(df_statmod['actual_price'],
                                    df_statmod['prediction_price'])))
`````
