```python
df = pd.read_csv("student_learning_activity_dataset.csv")
df
```

RUMUSAN MASALAH 

| Jenis | Ide Analisis | Variabel |
|---|---|---|
| Deskriptif | Berapa rata-rata jam belajar, kehadiran, jumlah tugas yang diselesaikan, dan nilai kuis mahasiswa? | `study_hours`, `attendance`, `assignments_completed`, `quiz_score` |
| Komparatif | Apakah mahasiswa dengan tingkat aktivitas belajar yang berbeda memiliki rata-rata nilai kuis yang berbeda? | `activity_level`, `quiz_score` |
| Hubungan | Apakah jumlah jam belajar berhubungan dengan nilai kuis mahasiswa? | `study_hours`, `quiz_score` |
| Hubungan | Apakah kehadiran berkaitan dengan performa? | `attendance`, `quiz_score` |
| Eksploratif | Bagaimana pola jam belajar mahasiswa pada setiap tingkat aktivitas (Low, Medium, dan High)? | `activity_level`, `quiz_score` |

```python
# RATA RATA JAM BELAJAR, KEHADIRAN, JUMLAH TUGAS YG SELESAI, DAN NILAI QUIZ
rata2 = df[[
    "study_hours",
    "attendance",
    "assignments_completed",
    "quiz_score"
]].mean()
print(round(rata2, 2))
```

```python
# PERBANDINGAN ANTARA TINGKAT AKTIFITAS DGN NILAI QUIZ
rata_quiz = df.groupby("activity_level")["quiz_score"].mean()

print(round(rata_quiz, 2))
```

```python
hubungan = df["study_hours"].corr(df["quiz_score"])

print("hubungan antara jam belajar dgn nilai kuiz : ", round(hubungan, 2))
print("terdapat hubungan positif antara waktu belajar dgn nilai quiz")
```

```python
hubungan2 = df["attendance"].corr(df["quiz_score"])
print(round(hubungan2, 2))
print("terdapat hubungan positif antara kehadiran dan nilai kuiz")
```

```python
rata_jam = df.groupby("activity_level")["quiz_score"].mean()

print(round(rata_jam, 2))
```

```python
def analyze_student(dataframe, student_id):
    try:
        mahasiswa = dataframe[dataframe["student_id"] == student_id]

        if mahasiswa.empty:
            print("mahasiswa tidak ditemukan")
            return

        data = mahasiswa.iloc[0]

        print("Student ID:", data["student_id"])
        print("Jam belajar:", data["study_hours"])
        print("Kehadiran:", data["attendance"])
        print("Nilai kuis:", data["quiz_score"])

        if data["quiz_score"] >= 75:
            print("Status: On Track")
            print("Rekomendasi: pertahanin belajarnya yaa!")
        else:
            print("Status: Needs Review")
            print("Rekomendasi: ayoo tingkatin lagi waktu belajarnya!")

    except:
        print("terjadi kesalahan")

analyze_student(df, 3)
```
